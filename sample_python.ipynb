Reverse a String using a Function

Check Palindrome using Function
Count Vowels in a String
Fibonacci Sequence (Recursion + Iterative)

```python
def revstr(a):
    return a[::-1]
a= str(input("Enter a string "))
print(revstr(a))
```

```python
def fact(n):
  if n==0 or n==1:
      return 1
  else:
      return n* fact(n-1)
n= int(input("Enter a no."))
print(fact(n))
```

```python
def pal(n):
  if n==n[::-1]:
     return 'palindrome'
  else:
     return 'not_palindrome'
n= str( input("enter a name "))
print(pal(n))

```

```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)
n= int( input("enter a number "))
print(fib(n))
```
