```python
import numpy as np
import pandas as pd
```

```python
rating= pd.read_csv('ratings.dat',names=['user_id','movie_id','rating','timestamp'],engine='python',delimiter='::',header= None,
                    encoding= 'latin1')
movie= pd.read_csv('movies.dat',names=['movie_id','title','genre'],engine='python',delimiter='::',header= None,
                    encoding= 'latin1')
```

```python
print(rating)
print(movie)
```

```python
ratingMatrix= np.ndarray(shape=(np.max(rating.movie_id.values),np.max(rating.user_id)),dtype=np.uint8)
ratingMatrix[rating.movie_id.values-1,rating.user_id.values-1]=rating.rating.values
print(ratingMatrix)
```

```python
normalizedMatrix=ratingMatrix-np.asarray([(np.mean(ratingMatrix,1))]).T
print(normalizedMatrix)
```

```python
a= normalizedMatrix.T/np.sqrt(ratingMatrix.shape[0]-1)
u,s,v = np.linalg.svd(a)
```

```python
def  similar (rating,movie_id,top_n):
  index= movie_id-1
  movie_row=rating[index,:]
  magnitude=np.sqrt(np.einsum('ij,ij -> i',rating,rating))
  similarity=np.dot(movie_row,rating.T)/(magnitude[index]*magnitude)
  sort_indexes = np.argsort(-similarity)
  return sort_indexes[: top_n]
```

```python
k=50
movie_id=318
top_n=10
sliced=v.T[:, :k]
indexes=similar(sliced,movie_id,top_n)

print('Recommendation for movie {0}:\n'.format(movie[movie.movie_id==movie_id].title.values[0]) )
for id in indexes +1:
  print(movie[movie.movie_id==id].title.values[0])
```
