```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ds =pd.read_csv('dataset.csv')
```

```python
print(ds.shape)
print(ds.head())
print(ds.describe())
```

```python
income= ds['INCOME'].values
spend= ds['SPEND'].values
x= np.array(list(zip(income,spend)))
x
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
  km= KMeans(n_clusters=i,random_state=0)
  km.fit(x)
  wcss.append(km.inertia_)
plt.plot(range(1,11),wcss,color='red',marker='8')
plt.title('Optimal K value')
plt.xlabel('No. of cluster')
plt.ylabel('WCSS')
plt.show()
```

```python
model= KMeans(n_clusters=4,random_state=0)
y_mean= model.fit_predict(x)
print(y_mean)
```

```python
plt.scatter(x[y_mean==0,0],x[y_mean==0,1],s=50,c='blue',label='1')
plt.scatter(x[y_mean==1,0],x[y_mean==1,1],s=50,c='black',label='2')
plt.scatter(x[y_mean==2,0],x[y_mean==2,1],s=50,c='red',label='3')
plt.scatter(x[y_mean==3,0],x[y_mean==3,1],s=50,c='green',label='4')
plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],s=100,marker='s',c='cyan',label='centroids')
plt.title('Income Spend Analsis')
plt.xlabel('Income')
plt.ylabel('Spend')
plt.legend()
plt.show()
```
