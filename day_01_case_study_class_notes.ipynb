```python
# 1) Data collection from source system.
import pandas as pd

customer_df = pd.read_csv('Mall_Customers.csv')
```

```python
# 2) Data Selection ( SELET Statement )

# SQL Query : SELECT * FROM customer_df

# Pandas Equvament:
customer_df
```

```python
# 3) Data Selection only with limited rows (LIMIT Statement)

# SQL Query mySQL : SELECT * FROM customer_df LIMIT 10 ;

# SQL Query Mssql : SELECT TOP 10 * FROM customer_df;

# Pandas Equvament:
customer_df.head(10)
```

```python
# 4) Data Selection with 10 rows of several attributes

# SQL Query : SELECT genre, Age FROM Customer_df LIMIT 10;

# Pandas Equvament:
customer_df[['Genre','Age']].head(10)
```

```python
# 5) Data Selection with Filter condition (SELECT .... WHERE )

# SQL Query : SELECT CustomerID,Genre FROM Customer_df WHERE Annual_Income_K_Dollars = 16;

# Pandas Equvament:
customer_df[customer_df['Annual_Income_K_Dollars'] == 16][['CustomerID','Genre']]
```

```python
# 6) Data Selection with mulitple filter condtitions ( SELECT WHERE(mulitple condition , all columns))

# SQL Query : SELECT * FROM Customer_df WHERE Annual_Income_K_Dollars >= 100 AND Genre ='Female';

# Pandas Equvament:

customer_df[(customer_df['Annual_Income_K_Dollars'] >= 100) & (customer_df['Genre'] == 'Female')]
```

```python
# 7) Data Selection with mulitple filter condtitions ( SELECT WHERE(mulitple condition , specific columns))

# SQL Query :  # SQL Query : SELECT Genre, Age, CustomerID FROM Customer_df WHERE Annual_Income_K_Dollars >= 100 AND Genre ='Female';

# Pandas Equvament:

customer_df[(customer_df['Annual_Income_K_Dollars'] >= 100) & (customer_df['Genre'] == 'Female')] [['Genre','Age','CustomerID']]
```

```python
customer_df.info()
```

```python
# 8) Summary Level Data (Aggregate level info)(avg, max, min)

# SQL Query :  SELECT AVG(Age), MAX(Age), MIN(Age) FROM Customer_df; alwasy have 1 record has output.

# Pandas Equvament:
# Normal way
customer_df['Age'].mean(), customer_df['Age'].max(), customer_df['Age'].min()

# Best Method
customer_df.agg({'Age':['mean','max','min','sum', 'count']})
```

```python
# 9) Data Selection with Unique value for the selected columns

# SQL Query : SELECT DISTINCT Genre FROM Customer_df;

# Pandas Equvament:

customer_df['Genre'].unique()
```

```python
# 10) Data Selection with unique value for mulitple columns

# SQL Query : SELECT DISTINCT Genre,Annual_Income_K_Dollars FROM customer_df;

# Pandas Equvament:
customer_df[['Genre','Annual_Income_K_Dollars']].drop_duplicates()
```

```python
# 11) Data Selection with unique value for mulitple columns in one vector (single column)

# SQL Query :

'''
SELECT DISTINCT Spending_Score_1_to_100 FROM customer_Df
UNION
SELECT DISTINCT Annual_Income_K_Dollars FROM customer_Df
'''
# Pandas Equvament:
import numpy as np

np.unique(customer_df[['Spending_Score_1_to_100','Annual_Income_K_Dollars']])
```

```python
# 12) Data Seletion with ordering the output (Ascending order)

# SQL Query : SELECT * FROM customer_df WHERE genre = 'Female' ORDER BY Spending_Score_1_to_100 ASC;

# Pandas Equvament:

customer_df[customer_df['Genre'] =='Female'].sort_values(by='Spending_Score_1_to_100',ascending=True).reset_index()
```

```python
# 13) Data Seletion with ordering the output (Desceding order)

# SQL Query : SELECT * FROM customer_df WHERE genre = 'Female' ORDER BY Spending_Score_1_to_100 DESC;

# Pandas Equvament: customer_df
customer_df[['Genre'] =='Female'].sort_values(by='Spending_Score_1_to_100',ascending=False).reset_index()
```
